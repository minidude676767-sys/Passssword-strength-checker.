```python
common = ["123456", "password", "qwerty", "admin", "iloveyou", "12345678"]

pw = input("Enter a password: ")
score = 0
tips = []

if len(pw) >= 8:
    score += 1
else:
    tips.append("Use at least 8 characters")

if any(c.islower() for c in pw) and any(c.isupper() for c in pw):
    score += 1
else:
    tips.append("Mix uppercase and lowercase letters")

if any(c.isdigit() for c in pw):
    score += 1
else:
    tips.append("Add a number")

if any(not c.isalnum() for c in pw):
    score += 1
else:
    tips.append("Add a symbol like @ or #")

if pw.lower() in common:
    score = 0
    tips.append("This is a very common password")

levels = ["Very weak", "Weak", "Fair", "Good", "Strong"]
print("Strength:", levels[score])
for t in tips:
    print("-", t)
```
